```python
import json

# Input file path
input_file = "merge_llama.json"  # Replace with your JSON file path
output_file = "cleaned_conflicts.json"

def clean_conflict_data(input_file, output_file):
    """
    Cleans and structures merge conflict data for LLM fine-tuning.
    """
    try:
        # Read the input JSON file
        with open(input_file, "r") as file:
            data = json.load(file)
        
        cleaned_data = []

        for item in data:
            conflict_resolution = item.get("conflict_resolution", "")
            if not conflict_resolution:
                continue  # Skip empty resolutions

            # Split the conflict into sections
            sections = conflict_resolution.split("<<<<<<<")
            for section in sections[1:]:  # Skip the first part (non-conflict)
                # Extract content between markers
                try:
                    parts = section.split(">>>>>>>")
                    conflict_body = parts[0].strip()  # Extract conflict details

                    # Split conflict body into versions
                    conflict_versions = conflict_body.split("=======")
                    version_a = conflict_versions[0].strip()
                    version_b = conflict_versions[1].strip() if len(conflict_versions) > 1 else ""

                    # Format into prompt-response structure
                    cleaned_data.append({
                        "prompt": f"Resolve the following merge conflict:\n\nVersion A:\n{version_a}\n\nVersion B:\n{version_b}\n\nProvide the resolved code:",
                        "response": f"{version_a}\n{version_b}"  # You can customize this if needed
                    })
                except Exception as e:
                    print(f"Error processing section: {section}\n{e}")

        # Save cleaned data to a new JSON file
        with open(output_file, "w") as outfile:
            json.dump(cleaned_data, outfile, indent=4)

        print(f"Cleaned data saved to {output_file}")

    except Exception as e:
        print(f"Error processing file: {e}")

# Run the cleaning process
clean_conflict_data(input_file, output_file)
```

```python
import json

# Input and output file paths
input_file = "merge_llama.json"  
output_file = "formatted_prompt_completion.jsonl"

def format_as_prompt_completion(input_file, output_file):
    """
    Formats merge conflict data into OpenAI's prompt-completion format.
    """
    try:
        # Load raw data
        with open(input_file, "r") as infile:
            data = json.load(infile)

        with open(output_file, "w") as outfile:
            for item in data:
                conflict_resolution = item.get("conflict_resolution", "")
                if not conflict_resolution:
                    continue

                # Split conflict into parts
                sections = conflict_resolution.split("<<<<<<<")
                for section in sections[1:]:
                    try:
                        parts = section.split(">>>>>>>")
                        conflict_body = parts[0].strip()
                        conflict_versions = conflict_body.split("=======")

                        version_a = conflict_versions[0].strip()
                        version_b = conflict_versions[1].strip() if len(conflict_versions) > 1 else ""

                        # Write as prompt-completion pairs
                        formatted_entry = {
                            "prompt": f"Resolve the following merge conflict:\n\nVersion A:\n{version_a}\n\nVersion B:\n{version_b}\n\nProvide the resolved code:",
                            "completion": f"{version_a}\n{version_b}"  # Customize the resolution logic as needed
                        }
                        outfile.write(json.dumps(formatted_entry) + "\n")
                    except Exception as e:
                        print(f"Error processing section: {section}\n{e}")
        print(f"Formatted data saved to {output_file}")

    except Exception as e:
        print(f"Error formatting data: {e}")

# Run the formatting function
format_as_prompt_completion(input_file, output_file)
```

```python
import json

# Input and output file paths
input_file = "merge_llama.json"
output_file = "formatted_chat_messages.jsonl"

def format_as_chat_messages(input_file, output_file):
    """
    Formats merge conflict data into OpenAI's chat message format.
    """
    try:
        # Load raw data
        with open(input_file, "r") as infile:
            data = json.load(infile)

        with open(output_file, "w") as outfile:
            for item in data:
                conflict_resolution = item.get("conflict_resolution", "")
                if not conflict_resolution:
                    continue

                # Split conflict into parts
                sections = conflict_resolution.split("<<<<<<<")
                for section in sections[1:]:
                    try:
                        parts = section.split(">>>>>>>")
                        conflict_body = parts[0].strip()
                        conflict_versions = conflict_body.split("=======")

                        version_a = conflict_versions[0].strip()
                        version_b = conflict_versions[1].strip() if len(conflict_versions) > 1 else ""

                        # Write as chat messages
                        formatted_entry = {
                            "messages": [
                                {"role": "system", "content": "You are an expert in resolving Git merge conflicts."},
                                {"role": "user", "content": f"Resolve the following conflict:\n\nVersion A:\n{version_a}\n\nVersion B:\n{version_b}"},
                                {"role": "assistant", "content": f"Resolved Code:\n{version_a}\n{version_b}"}  # Customize resolution as needed
                            ]
                        }
                        outfile.write(json.dumps(formatted_entry) + "\n")
                    except Exception as e:
                        print(f"Error processing section: {section}\n{e}")
        print(f"Formatted data saved to {output_file}")

    except Exception as e:
        print(f"Error formatting data: {e}")

# Run the formatting function
format_as_chat_messages(input_file, output_file)
```
